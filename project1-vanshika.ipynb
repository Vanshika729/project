```python
import torch
from torch import nn
from torch.utils.data import DataLoader, TensorDataset
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(device)

X_tr_normal_t = torch.tensor(X_tr_normal, dtype=torch.float32)
train_ds_ae = TensorDataset(X_tr_normal_t, X_tr_normal_t)
train_loader_ae = DataLoader(train_ds_ae, batch_size=256, shuffle=True)

class DenoisingAutoencoder(nn.Module):
    def __init__(self, input_dim):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Linear(input_dim, 64),
            nn.ReLU(),
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Linear(32, 16),
            nn.ReLU(),
            nn.Linear(16, 8),
            nn.ReLU()
        )
        self.decoder = nn.Sequential(
            nn.Linear(8, 16),
            nn.ReLU(),
            nn.Linear(16, 32),
            nn.ReLU(),
            nn.Linear(32, 64),
            nn.ReLU(),
            nn.Linear(64, input_dim)
        )

    def forward(self, x):
        z = self.encoder(x)
        return self.decoder(z)

ae = DenoisingAutoencoder(X_tr_normal.shape[1]).to(device)
print(ae)
```

```python
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(ae.parameters(), lr=1e-3)

def train_autoencoder(model, loader, epochs=30):
    model.train()
    for epoch in range(epochs):
        total_loss = 0
        for xb, yb in loader:
            xb = xb.to(device)
            optimizer.zero_grad()
            recon = model(xb)
            loss = criterion(recon, xb)
            loss.backward()
            optimizer.step()
            total_loss += loss.item() * xb.size(0)
        print(f"Epoch {epoch+1}/{epochs}, Loss: {total_loss/len(loader.dataset):.6f}")

train_autoencoder(ae, train_loader_ae, epochs=30)
```

```python
ae.eval()

with torch.no_grad():
    X_val_t = torch.tensor(X_val_scaled, dtype=torch.float32).to(device)
    test_X_t = torch.tensor(test_X_scaled, dtype=torch.float32).to(device)

    val_recon = ae(X_val_t).cpu().numpy()
    test_recon = ae(test_X_t).cpu().numpy()

val_err = np.mean((X_val_scaled - val_recon) ** 2, axis=1)
test_err = np.mean((test_X_scaled - test_recon) ** 2, axis=1)

print(val_err.shape, test_err.shape)
print(val_err[:5])
print(test_err[:5])
```

```python
from xgboost import XGBClassifier

xgb = XGBClassifier(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.1,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_lambda=1.0,
    random_state=42,
    eval_metric='logloss'
)

xgb.fit(X_tr_scaled, y_tr_np)

print("XGBoost trained")
```

```python
val_proba = xgb.predict_proba(X_val_scaled)[:, 1]
test_proba = xgb.predict_proba(test_X_scaled)[:, 1]

print(val_proba.shape, test_proba.shape)
print(val_proba[:5])
print(test_proba[:5])
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix

thresholds = np.arange(0.05, 0.96, 0.01)

best_thr = None
best_score = -1
best_metrics = None

for thr in thresholds:
    val_pred = (val_proba >= thr).astype(int)
    acc = accuracy_score(y_val_np, val_pred)
    tn, fp, fn, tp = confusion_matrix(y_val_np, val_pred).ravel()

    attack_recall = tp / (tp + fn) if (tp + fn) > 0 else 0
    normal_recall = tn / (tn + fp) if (tn + fp) > 0 else 0

    score = attack_recall + 0.10 * normal_recall

    if acc >= 0.90 and score > best_score:
        best_score = score
        best_thr = thr
        best_metrics = (acc, attack_recall, normal_recall, tn, fp, fn, tp)

print("Best threshold:", best_thr)
print("Best metrics:", best_metrics)
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.ensemble import RandomForestClassifier

print("="*70)
print("STAGE 3B: Model Training (XGB + LGB + RF Ensemble)")
print("="*70)

Xtr = np.asarray(X3b_tr_sel)
Xval = np.asarray(X3b_val_sel)
ytr = np.asarray(y3b_tr)
yval = np.asarray(y3b_val)

print(f"\nTraining data: {Xtr.shape}")
print(f"Validation data: {Xval.shape}")
print(f"Class distribution: {np.bincount(ytr)}")

# ============================================================================
# Model 1: XGBoost (with class weights for imbalance)
# ============================================================================
print("\n[1/3] Training XGBoost for Stage 3B...")

xgb3b = XGBClassifier(
    n_estimators=1000,
    max_depth=6,
    learning_rate=0.02,
    subsample=0.9,
    colsample_bytree=0.9,
    min_child_weight=1,
    gamma=0.5,
    reg_lambda=2.0,
    reg_alpha=0.1,
    objective="multi:softprob",
    eval_metric="mlogloss",
    random_state=42,
    n_jobs=-1,
    verbose=0
)

xgb3b.fit(Xtr, ytr)
print("  ✅ XGBoost trained!")

# ============================================================================
# Model 2: LightGBM (with is_unbalance for rare classes)
# ============================================================================
print("[2/3] Training LightGBM for Stage 3B...")

lgb3b = LGBMClassifier(
    n_estimators=1000,
    max_depth=10,
    learning_rate=0.02,
    num_leaves=50,
    subsample=0.9,
    colsample_bytree=0.9,
    is_unbalance=True,
    reg_lambda=2.0,
    reg_alpha=0.1,
    random_state=42,
    n_jobs=-1,
    verbose=-1
)

lgb3b.fit(Xtr, ytr)
print("  ✅ LightGBM trained!")

# ============================================================================
# Model 3: RandomForest (with class_weight for imbalance)
# ============================================================================
print("[3/3] Training RandomForest for Stage 3B...")

rf3b = RandomForestClassifier(
    n_estimators=800,
    max_depth=15,
    min_samples_leaf=2,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

rf3b.fit(Xtr, ytr)
print("  ✅ RandomForest trained!")

print("\n✅ All models trained!")

# ============================================================================
# Predictions: Soft Voting (Equal weights)
# ============================================================================
print("\nMaking ensemble predictions...")

xgb_proba = xgb3b.predict_proba(Xval)
lgb_proba = lgb3b.predict_proba(Xval)
rf_proba = rf3b.predict_proba(Xval)

# Equal voting
ensemble_proba = (xgb_proba + lgb_proba + rf_proba) / 3
pred_3b = np.argmax(ensemble_proba, axis=1)

# ============================================================================
# Evaluation
# ============================================================================
acc = accuracy_score(yval, pred_3b)
prec = precision_score(yval, pred_3b, average="macro", zero_division=0)
rec = recall_score(yval, pred_3b, average="macro", zero_division=0)
f1 = f1_score(yval, pred_3b, average="macro", zero_division=0)

print("\n" + "="*70)
print("=== STAGE 3B: Rare Attack Classification Results ===")
print("="*70)
print(f"Overall Accuracy: {acc:.4f}")
print(f"Macro Precision: {prec:.4f}")
print(f"Macro Recall: {rec:.4f}")
print(f"Macro F1: {f1:.4f}")

print(f"\nConfusion Matrix:")
cm = confusion_matrix(yval, pred_3b)
print(cm)

print(f"\nClassification Report:")
print(classification_report(yval, pred_3b, target_names=le3b.classes_, zero_division=0))

# ============================================================================
# Per-Class Breakdown
# ============================================================================
print("\n" + "="*70)
print("PER-CLASS DETAILED BREAKDOWN:")
print("="*70)

for idx, class_name in enumerate(le3b.classes_):
    support = np.sum(yval == idx)
    if support == 0:
        continue

    tp = cm[idx, idx]
    fp = np.sum(cm[:, idx]) - tp
    fn = np.sum(cm[idx, :]) - tp

    precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0
    f1_val = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

    print(f"\n{class_name}:")
    print(f"  Support: {support}")
    print(f"  True Positives: {tp}")
    print(f"  False Positives: {fp}")
    print(f"  False Negatives: {fn}")
    print(f"  Precision: {precision:.4f}")
    print(f"  Recall: {recall:.4f}")
    print(f"  F1-Score: {f1_val:.4f}")

print("\n✅ STAGE 3B Complete!")
```
